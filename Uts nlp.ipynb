```python
# Menghubungkan Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Import pandas
import pandas as pd

# Membaca file dari folder nlp
df = pd.read_csv('/content/drive/MyDrive/nlp/Ganjil.csv', sep=';', encoding='latin1')

# Menampilkan 5 data pertama
print(df.head())
```

```python
import re

def cleaning(text):
    text = str(text)

    # hapus URL
    text = re.sub(r'http\S+|www\S+', '', text)

    # hapus kata "sumber" dan metadata
    text = re.sub(r'sumber.*', '', text, flags=re.IGNORECASE)

    # hapus kata tidak penting (noise dari web)
    text = re.sub(r'membership|download|aplikasi', '', text, flags=re.IGNORECASE)

    # hapus angka
    text = re.sub(r'\d+', '', text)

    # hapus tanda baca & simbol
    text = re.sub(r'[^\w\s]', ' ', text)

    # hanya ambil huruf (biar bersih banget)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # ubah ke huruf kecil (langsung sekalian case folding)
    text = text.lower()

    # hapus spasi berlebih
    text = re.sub(r'\s+', ' ', text)

    # hapus spasi depan belakang
    text = text.strip()

    return text


# ==========================
# BUANG DATA KOSONG
# ==========================
df = df.dropna(subset=['Berita'])
df = df.reset_index(drop=True)

# ==========================
# TERAPKAN CLEANING
# ==========================
df['cleaning'] = df['Berita'].apply(cleaning)

# ==========================
# HAPUS HASIL KOSONG
# ==========================
df = df[df['cleaning'] != '']
df = df.reset_index(drop=True)

# ==========================
# CEK HASIL
# ==========================
print(df[['Berita', 'cleaning']].head())
```

```python
# install & import
!pip install nltk

import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords

# stopword bahasa indonesia
stop_words = set(stopwords.words('indonesian'))

# tambahan stopword (khusus dataset kamu)
custom_stopwords = {
    'kompas', 'co', 'id', 'kpk', 'jakarta'
}

# gabungkan
stop_words = stop_words.union(custom_stopwords)

# fungsi stopword removal
def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

# terapkan ke normalization
df['stopword'] = df['normalization'].apply(remove_stopwords)

# ==========================
# OUTPUT TANPA INDEX
# ==========================
print("Tanpa index:")
for s in df['stopword'].head():
    print(s)
```
