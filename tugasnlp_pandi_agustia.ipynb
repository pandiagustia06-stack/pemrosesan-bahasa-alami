```python
import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from nltk.corpus import stopwords
from collections import Counter
from google.colab import drive
drive.mount('/content/drive')
nltk.download('stopwords')
df = pd.read_csv('/content/drive/MyDrive/dataset/puisi.csv')
```

```python
text = " ".join(df['puisi'].astype(str))
```

```python
text = text.lower()
text = re.sub(r'[^a-zA-Z\s]', '', text)
```

```python
stop_words = set(stopwords.words('indonesian'))
```

```python
words = text.split()
```

```python
filtered_words = [word for word in words if word not in stop_words]
```

```python
word_counts = Counter(filtered_words)
```

```python
top_words = word_counts.most_common(15)
```

```python
kata = [item[0] for item in top_words]
jumlah = [item[1] for item in top_words]
```

```python
plt.figure(figsize=(10,6))
plt.bar(kata, jumlah)
plt.xlabel('Kata')
plt.ylabel('Frekuensi')
plt.title('Frekuensi Kata Puisi')
plt.xticks(rotation=45)
plt.show()
```

```python
# WordCloud
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(" ".join(filtered_words))
```

```python
plt.figure(figsize=(12,6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('WordCloud Puisi')
plt.show()
```
